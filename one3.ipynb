```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import torch

# Load your dataset
df = pd.read_csv("/Users/abhisheksingh/Downloads/civic_issues_dataset_8k_plus.csv")

print(df.head())
print(df['department'].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Drop rows with missing descriptions
df = df[df['description'].notna()]

X = df['description']
y = df['department']

# Encode department labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

print("Classes:", label_encoder.classes_)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report

# Build pipeline
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(ngram_range=(1,2), max_features=20000)),
    ('clf', LogisticRegression(max_iter=500))
])

# Train
pipeline.fit(X_train, y_train)

# Evaluate
y_pred = pipeline.predict(X_test)
print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))
```

```python
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments
import torch
from torch.utils.data import Dataset

# Prepare tokenizer
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

class CivicDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts[idx])
        label = self.labels[idx]
        encoding = self.tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=self.max_len,
            return_tensors="pt"
        )
        return {
            "input_ids": encoding["input_ids"].flatten(),
            "attention_mask": encoding["attention_mask"].flatten(),
            "labels": torch.tensor(label, dtype=torch.long)
        }

train_dataset = CivicDataset(X_train.tolist(), y_train, tokenizer)
test_dataset  = CivicDataset(X_test.tolist(), y_test, tokenizer)

# Load model
model = BertForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=len(label_encoder.classes_))

# Training arguments
from transformers import TrainingArguments, Trainer

training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",   # ✅ use this instead
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01
)


trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset
)

trainer.train()
```

```python
results = trainer.evaluate(test_dataset)
print(results)
```

```python
import numpy as np
from sklearn.metrics import classification_report

preds_output = trainer.predict(test_dataset)
y_true = preds_output.label_ids
y_pred = np.argmax(preds_output.predictions, axis=1)

print(classification_report(y_true, y_pred, target_names=label_encoder.classes_))
```

```python
from transformers import pipeline

classifier = pipeline("text-classification", model=model, tokenizer=tokenizer)

print(classifier("The streetlights are not working on my road."))
print(classifier("There is sewage overflow near my house."))
print(classifier("The power supply is down since yesterday."))
```

```python
# Create mapping from label encoder
id2label = {i: label for i, label in enumerate(label_encoder.classes_)}
label2id = {label: i for i, label in enumerate(label_encoder.classes_)}

# Reload model with mapping
model = BertForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=len(label_encoder.classes_),
    id2label=id2label,
    label2id=label2id
)

# Recreate pipeline
classifier = pipeline("text-classification", model=model, tokenizer=tokenizer)

print(classifier("The streetlights are not working on my road."))
print(classifier("There is sewage overflow near my house."))
print(classifier("The power supply is down since yesterday."))
```

```python
from transformers import BertForSequenceClassification, BertTokenizer, pipeline

# reload fine-tuned model
model = BertForSequenceClassification.from_pretrained(
    "./dept_model",
    num_labels=11,
    id2label={
        0: 'Animal Issues',
        1: 'Electricity/Streetlight',
        2: 'Flood/Waterlogging',
        3: 'Health',
        4: 'Noise Pollution',
        5: 'Protest/Traffic',
        6: 'Roads',
        7: 'Sanitation',
        8: 'Unknown',
        9: 'Urban Development',
        10: 'Water Supply'
    },
    label2id={
        'Animal Issues': 0,
        'Electricity/Streetlight': 1,
        'Flood/Waterlogging': 2,
        'Health': 3,
        'Noise Pollution': 4,
        'Protest/Traffic': 5,
        'Roads': 6,
        'Sanitation': 7,
        'Unknown': 8,
        'Urban Development': 9,
        'Water Supply': 10
    }
)

tokenizer = BertTokenizer.from_pretrained("./dept_model")

classifier = pipeline("text-classification", model=model, tokenizer=tokenizer)

print(classifier("The streetlights are broken on the main road."))
```

```python
def predict_department(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=128)
    outputs = model(**inputs)
    pred = torch.argmax(outputs.logits, dim=1).item()
    return label_encoder.inverse_transform([pred])[0]

print(predict_department("Streetlight not working near my house"))
print(predict_department("पानी की सप्लाई पिछले 3 दिनों से बंद है"))
```
